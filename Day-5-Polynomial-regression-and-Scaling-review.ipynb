```python
((df2.x1-df2.x1.mean())/df2.x1.std()).describe()
```

```python
((df2.x1-df2.x1.min())/(df2.x1.max()-df2.x1.min())).describe()
```

```python
def normalize(feature):
    return (df2[feature]-df2[feature].min())/(df2[feature].max()-df2[feature].min())
    
```

```python
normalize("x3")
```

```python
df2["x1_norm"],df2["x2_norm"],df2["x3_norm"]=normalize("x1"),normalize("x2"),normalize("x3")
```

```python
for column in ["x1","x2","x3"]:
    df2[f"{column}_norm"]=normalize(column)
```

```python
X_2=df2[["x0","x1_norm","x2_norm","x3_norm"]].to_numpy()
```
