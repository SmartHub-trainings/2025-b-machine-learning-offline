```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_breast_cancer
```

```python
X,y=load_breast_cancer(return_X_y=True)
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y)
```

```python
knn_model = KNeighborsClassifier()
knn_model.fit(X_train,y_train)
training_score=knn_model.score(X_train,y_train)
testing_score=knn_model.score(X_test,y_test)
print(f"Training Score : {training_score}")
print(f"Testing Score : {testing_score}")
```

```python
knn_model = KNeighborsClassifier(1)
knn_model.fit(X_train,y_train)
training_score=knn_model.score(X_train,y_train)
testing_score=knn_model.score(X_test,y_test)
print(f"Training Score : {training_score}")
print(f"Testing Score : {testing_score}")
```

```python
import matplotlib.pyplot as plt
testing_scores=[]
training_scores=[]
neihb=range(3,31)
for k in neihb:
    knn_model = KNeighborsClassifier(k)
    knn_model.fit(X_train,y_train)
    training_score=knn_model.score(X_train,y_train)
    testing_score=knn_model.score(X_test,y_test)
    testing_scores.append(testing_score)
    training_scores.append(training_score)
# print(f"Training scores: {training_scores}")
# print(f"Testing scores: {testing_scores}")
plt.plot(neihb,testing_scores,label="Testing Scores")
plt.plot(neihb,training_scores,label="Training Scores")
plt.legend()
plt.xlabel("Number of neihbors")
plt.ylabel("Accuracy")
plt.grid()
```

```python
import matplotlib.pyplot as plt
testing_scores=[]
training_scores=[]
neihb=range(3,31)
for k in neihb:
    knn_model = KNeighborsClassifier(k,p=5)
    knn_model.fit(X_train,y_train)
    training_score=knn_model.score(X_train,y_train)
    testing_score=knn_model.score(X_test,y_test)
    testing_scores.append(testing_score)
    training_scores.append(training_score)
# print(f"Training scores: {training_scores}")
# print(f"Testing scores: {testing_scores}")
plt.plot(neihb,testing_scores,label="Testing Scores")
plt.plot(neihb,training_scores,label="Training Scores")
plt.legend()
plt.xlabel("Number of neihbors")
plt.ylabel("Accuracy")
plt.grid()
```
