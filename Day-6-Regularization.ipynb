```python
X_train,y_train=df[df.split=="train"].x.values.reshape(-1,1),df[df.split=="train"].y.values
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
sim_linear_model=LinearRegression()
sim_linear_model.fit(X_train,y_train)
training_score=sim_linear_model.score(X_train,y_train)
testing_score=sim_linear_model.score(X_test,y_test)
print(f"Training accuracy: {training_score}")
print(f"Testing accuracy: {testing_score}")
```

```python
polyfeature=PolynomialFeatures(degree=3)
X_train_poly=polyfeature.fit_transform(X_train)
X_test_poly=polyfeature.fit_transform(X_test)
multi_linear_model3=LinearRegression()
multi_linear_model3.fit(X_train_poly,y_train)
training_score=multi_linear_model3.score(X_train_poly,y_train)
testing_score=multi_linear_model3.score(X_test_poly,y_test)
print(f"Training accuracy: {training_score}")
print(f"Testing accuracy: {testing_score}")
```

```python
polyfeature=PolynomialFeatures(degree=15)
X_train_poly=polyfeature.fit_transform(X_train)
X_test_poly=polyfeature.fit_transform(X_test)
multi_linear_model15=LinearRegression()
multi_linear_model15.fit(X_train_poly,y_train)
training_score=multi_linear_model15.score(X_train_poly,y_train)
testing_score=multi_linear_model15.score(X_test_poly,y_test)
print(f"Training accuracy: {training_score}")
print(f"Testing accuracy: {testing_score}")
```

## Using ridge Regularization

```python
from sklearn.linear_model import Ridge,Lasso,ElasticNet
```

```python
ridge_model15=Ridge()
ridge_model15.fit(X_train_poly,y_train)
training_score=ridge_model15.score(X_train_poly,y_train)
testing_score=ridge_model15.score(X_test_poly,y_test)
print(f"Training accuracy: {training_score}")
print(f"Testing accuracy: {testing_score}")
print(ridge_model15.coef_)
print(ridge_model15.intercept_)
```

```python
lasso_model15=Lasso(alpha=0.9)
lasso_model15.fit(X_train_poly,y_train)
training_score=lasso_model15.score(X_train_poly,y_train)
testing_score=lasso_model15.score(X_test_poly,y_test)
print(f"Training accuracy: {training_score}")
print(f"Testing accuracy: {testing_score}")
print(lasso_model15.coef_)
print(lasso_model15.intercept_)
```

```python
import numpy as np
```

```python
alpha=0.001
test_scores=[]
train_scores=[]
alphas=[]
for i in range(50):
    elastic_model15=ElasticNet(alpha)
    elastic_model15.fit(X_train_poly,y_train)
    training_score=elastic_model15.score(X_train_poly,y_train)
    testing_score=elastic_model15.score(X_test_poly,y_test)
    test_scores.append(testing_score)
    train_scores.append(training_score)
    # print(f"Training accuracy: {training_score}")
    # print(f"Testing accuracy: {testing_score}")
    # print(elastic_model15.coef_)
    # print(elastic_model15.intercept_)
    alphas.append(alpha)
    alpha =alpha+0.001
test_scores=np.array(test_scores)
train_scores=np.array(train_scores)
plt.plot(alphas,test_scores)
plt.plot(alphas,train_scores)
```
