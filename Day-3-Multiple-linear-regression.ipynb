```python
from sklearn.metrics import mean_squared_error
```

## Simple model for each feature

```python

for i  in range(8):
    linear_model= LinearRegression()
    X=data[:,i].reshape(-1,1)
    y=housing.target
    linear_model.fit(X,y)
    score = linear_model.score(X,y)
    y_pred = linear_model.predict(X)
    corr_cor=np.corrcoef(X.flatten(),y)[0,1]
    rmse=(mean_squared_error(y,y_pred))**0.5
    print(f"For feature '{housing.feature_names[i]}'")
    print(f"Correlation Coefficient: {corr_cor}")
    print(f"Goodness of fit : {score*100}")
    print(f"Root mean Squared Error: {rmse}")
    print("="*30)

```
