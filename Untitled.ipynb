```python
import numpy as np
```

```python
x=np.array([3,7,2,1,4])
y=np.array([6,15,4,3,9])
```

```python
n=len(x)
sum_x= x.sum()
sum_y=y.sum()
sum_xy= (x*y).sum()
sum_x_sq= (x*x).sum()
```

```python
sum_x
```

```python
sum_y
```

```python
x*y
```

```python
sum_xy
```

```python
beta_1= (n*sum_xy-sum_x*sum_y)/(n*sum_x_sq-sum_x**2)
```

```python
beta_1
```

```python
beta_0= y.mean()-beta_1*x.mean()
```

```python
beta_0
```

```python
def estimated_y(x):
    return beta_0+beta_1*x
```

```python
estimated_y(3)
```

```python
y_pred=[]
for i in x:
    y_pred.append(estimated_y(i))
y_pred=np.array(y_pred)
```
